```python
import requests

url = "https://wellhub.com/en-us/search/tx/dallas/"

try:
    response = requests.get(url)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
    print(f"Status Code: {response.status_code}")
    print("Response content snippet:")
    print(response.text[:500]) # Print first 500 characters of the response
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

```python
from bs4 import BeautifulSoup

# Assuming 'response' object is available from the previous cell execution
# If not, you might need to re-run the request:
# response = requests.get(url)
# response.raise_for_status()

soup = BeautifulSoup(response.text, 'html.parser')

selector_content = soup.select_one("body > script:nth-child(20)")

if selector_content:
    print("Content from selector 'body > script:nth-child(20)':")
    print(selector_content.prettify())
else:
    print("Selector 'body > script:nth-child(20)' not found in the HTML.")
```
