```python
import requests
import json
import os
import time
import random
from datetime import datetime, timezone

# ============================================================
# CONFIG
# ============================================================

URL = (
    "https://marinemaxproduction144uc8r3u.org.coveo.com/"
    "rest/organizations/marinemaxproduction144uc8r3u/"
    "commerce/v2/search"
)

OUTPUT_DIR = "marinemax_json"
PER_PAGE = 32

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ============================================================
# ZIP CODES
# Excluding PO BOX and UNIQUE ZIP codes
# ============================================================

ZIPCODES = [
    "68102",
    "68104",
    "68105",
    "68106",
    "68107",
    "68108",
    "68110",
    "68111",
    "68112",
    "68113",
    "68114",
    "68116",
    "68117",
    "68118",
    "68122",
    "68124",
    "68127",
    "68130",
    "68131",
    "68132",
    "68134",
    "68135",
    "68136",
    "68137",
    "68138",
    "68142",
    "68144",
    "68147",
    "68152",
    "68154",
    "68157",
    "68164",
    "68176",
]


# ============================================================
# HEADERS
# ============================================================

HEADERS = {
    "accept": "*/*",
    "accept-encoding": "gzip, deflate, br, zstd",
    "accept-language": "en-US,en;q=0.9,id;q=0.8,fa;q=0.7",

    "authorization": (
        "Bearer eyJhbGciOiJIUzI1NiJ9."
        "eyJ2OCI6dHJ1ZSwidG9rZW5JZCI6InhsZ3ZtNXI1ank0cTd1aHNkeGRlZmFzaDJ5Iiw"
        "ib3JnYW5pemF0aW9uIjoibWFyaW5lbWF4cHJvZHVjdGlvbjE0NHVjOHIzdSIsInVzZXJJ"
        "ZHMiOlt7InR5cGUiOiJVc2VyIiwibmFtZSI6ImFub255bW91cyIsInByb3ZpZGVyIjoiR"
        "W1haWwgU2VjdXJpdHkgUHJvdmlkZXIifV0sInJvbGVzIjpbInF1ZXJ5RXhlY3V0b3IiXS"
        "wiaXNzIjoiU2VhcmNoQXBpIiwiZXhwIjoxNzkwNzcxMzc0LCJpYXQiOjE3OTA2ODQ5NzR"
        "9.vyGouwBttDTmopmSmbZLuUHEtDV4NkXJ6m5WU5Cq1wc"
    ),

    "content-type": "application/json",
    "origin": "https://www.marinemax.com",
    "referer": "https://www.marinemax.com/",

    "sec-ch-ua": (
        "\"Chromium\";v=\"154\", "
        "\"Microsoft Edge\";v=\"154\", "
        "\"Not A(Brand\";v=\"99\""
    ),

    "sec-ch-ua-mobile": "?0",
    "sec-ch-ua-platform": "\"Windows\"",
    "sec-fetch-dest": "empty",
    "sec-fetch-mode": "cors",
    "sec-fetch-site": "cross-site",

    "user-agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/154.0.0.0 Safari/537.36 "
        "Edg/154.0.0.0"
    )
}


# ============================================================
# SESSION
# ============================================================

session = requests.Session()
session.headers.update(HEADERS)


# ============================================================
# BUILD PAYLOAD
# ============================================================

def build_payload(zipcode, page):

    timestamp = datetime.now(timezone.utc).isoformat()

    payload = {
        "trackingId": "market_31936807149",
        "clientId": "50852086-c274-4dd0-a021-1ea72f264ed5",

        "context": {
            "user": {
                "userAgent": HEADERS["user-agent"]
            },

            "view": {
                "url": "https://www.marinemax.com/search"
            },

            "capture": True,
            "cart": [],

            "source": [
                "@coveo/headless@3.57.0"
            ],

            # Keeping your original coordinates.
            # The ZIP itself changes below.
            "location": {
                "latitude": 41.2919,
                "longitude": -95.9999
            },

            "location_type": "zip",
            "location_lat": 41.2919,
            "location_lng": -95.9999,
            "location_zip": zipcode,
            "location_timestamp": timestamp,
            "location_precision": "approx",
            "location_radius_miles": 500,

            "custom": {
                "location_type": "zip",
                "location_lat": 41.2919,
                "location_lng": -95.9999,
                "location_zip": zipcode,
                "location_timestamp": timestamp,
                "location_precision": "approx",
                "location_radius_miles": 500
            }
        },

        "language": "en",
        "country": "US",
        "currency": "USD",

        "page": page,
        "perPage": PER_PAGE,

        "facets": [],

        "sort": {
            "sortCriteria": "relevance"
        },

        "query": "",

        "enableResults": False,

        "numberOfResults": PER_PAGE,
        "pageSize": PER_PAGE
    }

    return payload


# ============================================================
# REQUEST WITH RETRIES
# ============================================================

def fetch_page(zipcode, page, max_retries=3):

    payload = build_payload(zipcode, page)

    for attempt in range(1, max_retries + 1):

        try:

            response = session.post(
                URL,
                json=payload,
                timeout=60
            )

            print(
                f"ZIP {zipcode} | "
                f"Page {page} | "
                f"HTTP {response.status_code}"
            )

            response.raise_for_status()

            return response.json()

        except requests.RequestException as e:

            print(
                f"Request error | "
                f"ZIP {zipcode} | "
                f"Page {page} | "
                f"Attempt {attempt}/{max_retries}"
            )

            print(e)

        except json.JSONDecodeError as e:

            print(
                f"JSON decode error | "
                f"ZIP {zipcode} | "
                f"Page {page}"
            )

            print(e)

        if attempt < max_retries:

            sleep_time = attempt * 5

            print(
                f"Retrying in {sleep_time} seconds..."
            )

            time.sleep(sleep_time)

    return None


# ============================================================
# SCRAPE ONE ZIP CODE
# ============================================================

def scrape_zipcode(zipcode):

    output_file = os.path.join(
        OUTPUT_DIR,
        f"{zipcode}.json"
    )

    # --------------------------------------------------------
    # Resume support
    # --------------------------------------------------------

    if os.path.exists(output_file):

        print(
            f"\nZIP {zipcode} already exists. "
            f"Skipping."
        )

        return

    print("\n" + "=" * 70)
    print(f"Starting ZIP: {zipcode}")
    print("=" * 70)

    page = 0
    all_products = []

    while True:

        data = fetch_page(
            zipcode=zipcode,
            page=page
        )

        if data is None:

            print(
                f"Failed ZIP {zipcode}, page {page}. "
                f"Stopping this ZIP."
            )

            break

        products = data.get("products", [])

        result_count = len(products)

        print(
            f"ZIP {zipcode} | "
            f"Page {page} | "
            f"Products: {result_count}"
        )

        # Add products from current page
        all_products.extend(products)

        # ----------------------------------------------------
        # Last page
        # ----------------------------------------------------

        if result_count < PER_PAGE:

            print(
                f"Last page found for ZIP {zipcode}: "
                f"page {page}"
            )

            break

        # Next page
        page += 1

        # Small polite delay
        delay = random.uniform(1.5, 3.0)

        time.sleep(delay)

    # --------------------------------------------------------
    # Save ZIP result
    # --------------------------------------------------------

    output_data = {
        "zipcode": zipcode,
        "total_products": len(all_products),
        "total_pages": page + 1,
        "scraped_at": datetime.now(timezone.utc).isoformat(),
        "products": all_products
    }

    with open(
        output_file,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            output_data,
            f,
            ensure_ascii=False,
            indent=2
        )

    print(
        f"Saved ZIP {zipcode}: "
        f"{len(all_products)} products"
    )

    print(
        f"File: {output_file}"
    )


# ============================================================
# MAIN
# ============================================================

def main():

    print(
        f"Total ZIP codes: {len(ZIPCODES)}"
    )

    for index, zipcode in enumerate(
        ZIPCODES,
        start=1
    ):

        print(
            f"\nZIP {index}/{len(ZIPCODES)}"
        )

        try:

            scrape_zipcode(zipcode)

        except KeyboardInterrupt:

            print("\nStopped manually.")

            break

        except Exception as e:

            print(
                f"Unexpected error for ZIP "
                f"{zipcode}: {e}"
            )

            # Continue with next ZIP
            continue

        # Delay before next ZIP
        time.sleep(
            random.uniform(2.0, 4.0)
        )

    print("\nAll ZIP codes completed.")


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    main()
```

```python
import os
import json

# ============================================================
# CONFIG
# ============================================================

INPUT_DIR = "marinemax_json"

ALL_OUTPUT_FILE = "marinemax_all_products.json"
UNIQUE_OUTPUT_FILE = "marinemax_unique_products.json"


# ============================================================
# MAIN
# ============================================================

def main():

    all_products = []

    # ec_product_id -> product
    unique_products = {}

    files = sorted([
        filename
        for filename in os.listdir(INPUT_DIR)
        if filename.lower().endswith(".json")
    ])

    print(f"Found {len(files)} JSON files.")
    print("=" * 70)

    for index, filename in enumerate(files, start=1):

        filepath = os.path.join(
            INPUT_DIR,
            filename
        )

        try:

            with open(
                filepath,
                "r",
                encoding="utf-8"
            ) as f:
                data = json.load(f)

            products = data.get("products", [])

            print(
                f"[{index}/{len(files)}] "
                f"{filename}: {len(products)} products"
            )

            # ------------------------------------------------
            # Add ALL products
            # ------------------------------------------------

            all_products.extend(products)

            # ------------------------------------------------
            # Add UNIQUE products
            # ------------------------------------------------

            for product in products:

                product_id = product.get("ec_product_id")

                if product_id:

                    # Keep first occurrence
                    if product_id not in unique_products:
                        unique_products[product_id] = product

                else:
                    print(
                        f"  WARNING: product without "
                        f"ec_product_id: "
                        f"{product.get('ec_name', 'Unknown')}"
                    )

        except json.JSONDecodeError as e:

            print(
                f"ERROR reading {filename}: "
                f"Invalid JSON - {e}"
            )

        except Exception as e:

            print(
                f"ERROR reading {filename}: {e}"
            )

    # ========================================================
    # CONVERT UNIQUE DICTIONARY TO LIST
    # ========================================================

    unique_products_list = list(
        unique_products.values()
    )


    # ========================================================
    # SAVE ALL PRODUCTS
    # ========================================================

    all_output = {
        "total_products": len(all_products),
        "products": all_products
    }

    with open(
        ALL_OUTPUT_FILE,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            all_output,
            f,
            ensure_ascii=False,
            indent=2
        )


    # ========================================================
    # SAVE UNIQUE PRODUCTS
    # ========================================================

    unique_output = {
        "total_products": len(unique_products_list),
        "unique_by": "ec_product_id",
        "products": unique_products_list
    }

    with open(
        UNIQUE_OUTPUT_FILE,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            unique_output,
            f,
            ensure_ascii=False,
            indent=2
        )


    # ========================================================
    # SUMMARY
    # ========================================================

    duplicates = (
        len(all_products)
        - len(unique_products_list)
    )

    print("\n" + "=" * 70)
    print("COMPLETED")
    print("=" * 70)

    print(
        f"JSON files processed : {len(files):,}"
    )

    print(
        f"All records          : {len(all_products):,}"
    )

    print(
        f"Unique records       : {len(unique_products_list):,}"
    )

    print(
        f"Duplicate records    : {duplicates:,}"
    )

    print("\nOutput files:")

    print(
        f"  All    -> {ALL_OUTPUT_FILE}"
    )

    print(
        f"  Unique -> {UNIQUE_OUTPUT_FILE}"
    )


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    main()
```
